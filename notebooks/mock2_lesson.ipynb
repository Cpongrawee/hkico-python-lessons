Analyze the given below code?

```python
print("Recursive Function")
def factorial(n):
    return(n*factorial(n-1))
factorial(4)
```

- A. Recursive Function 24.
- B. Recursive Function
- C. Function runs infinitely and causes a StackOverflowError.

### 📝 คำตอบของฉัน (ห้ามรันโค้ดก่อนตอบ!)
**ดับเบิลคลิกที่ cell นี้เพื่อพิมพ์**

- คำตอบ:

````python
print("Recursive Function")
def factorial(n):
    return(n*factorial(n-1))
````

````python
print("Recursive Function")
def factorial(n):
    return(n*factorial(n-1))
factorial(4)
````

<details>
<summary>✅ เฉลยข้อ 3 (เปิดเมื่อครูบอก)</summary>

**คำตอบ: C**

`factorial` เรียกตัวเองไปเรื่อย ๆ: 4, 3, 2, 1, 0, −1, −2, … เพราะ**ไม่มี base case** ให้หยุด
Python จะหยุดเองเมื่อเรียกลึกเกินประมาณ 1000 ชั้น แล้วแจ้ง `RecursionError` ซึ่งก็คือ "stack overflow" ในภาษาอื่น (ชื่อ `StackOverflowError` มาจาก Java)
- บรรทัด `Recursive Function` จะพิมพ์ออกมาก่อนเสมอ แต่ไม่มีตัวเลือกไหนพูดถึงทั้งสองอย่าง C จึงใกล้ที่สุด
- A ✘ ต่อให้มี base case ก็ไม่มี `print` ผลของ `factorial(4)` (ในไฟล์ .py ค่าที่ไม่ได้ print จะไม่แสดง)
- B ✘ โปรแกรมไม่ได้จบเงียบ ๆ แต่พังด้วย error
- D ✘ `return(...)` ใส่วงเล็บได้ เหมือน `return ...` ทุกอย่าง

</details>

### 📚 ความรู้ที่ต้องรู้

**1) recursion ต้องมี base case**: เงื่อนไขที่ "ไม่เรียกตัวเองอีก" ถ้าไม่มี จะเรียกต่อไปไม่สิ้นสุด

```python
def factorial(n):
    if n == 0:        # base case
        return 1
    return n * factorial(n - 1)
```

**2) Python จำกัดความลึก**: ประมาณ 1000 ชั้น เกินแล้วได้ `RecursionError: maximum recursion depth exceeded`

**3) โค้ดก่อนหน้าที่ error ทำงานไปแล้ว**: `print` บรรทัดแรกพิมพ์ออกมาก่อนที่ `factorial` จะพัง

## ข้อ 5 — ⭐ `*args` รับ keyword argument ไม่ได้

What is the output of the following `displayPerson()` function call?

```python
def displayPerson(*args):
    for i in args:
        print(i)

displayPerson(name="Emma", age="25")
```

- A. Emma<br>25
- B. name<br>age
- C. 25
- D. TypeError
- E. None of the above

### 📝 คำตอบของฉัน (ห้ามรันโค้ดก่อนตอบ!)
**ดับเบิลคลิกที่ cell นี้เพื่อพิมพ์**

- คำตอบ: `D`
- ตัดตัวเลือกไหนทิ้ง เพราะอะไร:

````python
(1,3 )
````

````python
(1,3) == (3,1)
````

````python
{1,3} == {1,3, 1, 3}
````

<details>
<summary>✅ เฉลยข้อ 5 (เปิดเมื่อครูบอก)</summary>

**คำตอบ: D. TypeError**

`*args` เก็บได้แค่ argument แบบ**ตามตำแหน่ง** (positional) แต่ `name="Emma"` เป็นแบบ**ระบุชื่อ** (keyword)
→ `TypeError: displayPerson() got an unexpected keyword argument 'name'`
- A ✘ จะได้ถ้าเรียกแบบ `displayPerson("Emma", "25")` (ดูทดลอง 5.1)
- B ✘ จะได้ถ้าประกาศเป็น `**kwargs` แล้ววนลูป (ดูทดลอง 5.3)
- C ✘ ไม่มีทางได้แค่ค่าสุดท้าย

</details>

### 📚 ความรู้ที่ต้องรู้

**1) ส่ง argument ได้ 2 แบบ**
- ตามตำแหน่ง (positional): `f("Emma", "25")`
- ระบุชื่อ (keyword): `f(name="Emma", age="25")`

**2) `*` กับ `**` ในการประกาศฟังก์ชัน**

| ประกาศ | รับอะไร | เก็บเป็น |
|---|---|
| `def f(*args)` | positional กี่ตัวก็ได้ | tuple |
| `def f(**kwargs)` | keyword กี่ตัวก็ได้ | dict |

ชื่อ `args`/`kwargs` เปลี่ยนได้ สิ่งสำคัญคือ `*` กับ `**`

**3) ⚠️ ส่งแบบที่ฟังก์ชันไม่รับ → `TypeError`**

## ข้อ 19 — ⭐ `x = y += 2` เขียนไม่ได้

What is the output of the following assignment operator?

```python
y = 10
x = y += 2
print(x)
```

- A. 10
- B. 12
- C. 1
- D. 2
- E. Syntax Error

### 📝 คำตอบของฉัน (ห้ามรันโค้ดก่อนตอบ!)
**ดับเบิลคลิกที่ cell นี้เพื่อพิมพ์**

- คำตอบ:
- ตัดตัวเลือกไหนทิ้ง เพราะอะไร:

<details>
<summary>✅ เฉลยข้อ 19 (เปิดเมื่อครูบอก)</summary>

**คำตอบ: E. Syntax Error**

`y += 2` เป็น**คำสั่ง** (statement) ไม่ใช่ค่า จึงเอาไปใส่ทางขวาของ `x =` ไม่ได้ Python ตรวจเจอตั้งแต่ก่อนรัน → `SyntaxError`
- B ✘ จะได้ถ้าแยกเป็น 2 บรรทัด `y += 2` แล้ว `x = y` (ดูทดลอง 19.1)
- A ✘ คิดว่า `x` ได้ค่า `y` ก่อนบวก

</details>

### 📚 ความรู้ที่ต้องรู้

**1) `y += 2` คือ `y = y + 2`**: เป็นคำสั่งเปลี่ยนค่า ไม่ได้ "ให้ค่า" ออกมา

**2) `=` ต่อกันหลายตัวได้ แต่ `+=` ไม่ได้**

| โค้ด | ผล |
|---|---|
| `x = y = 12` | ✅ ทั้งสองตัวได้ 12 |
| `x = y += 2` | ❌ `SyntaxError` |

**3) `SyntaxError` เกิดก่อนรัน**: บรรทัด `y = 10` ก็ไม่ได้ทำงาน ต่างจาก error อื่นที่ทำงานไปถึงบรรทัดที่พัง (เทียบกับข้อ 3)

## ข้อ 20 — if ซ้อน if

Given the nested if-else below, what will `x` become after the statement is executed?

````python
x = 0
a = 5
b = 5
if(a > 0):
  if(b < 0):
      x = x+5
  elif(a>5):
      x = x+4
  else:
      x = x+3
else:
  x =x+2
print(x)
````

- A. 0
- B. 4
- C. 2
- D. 3
- E. None of the above

### 📝 คำตอบของฉัน (ห้ามรันโค้ดก่อนตอบ!)
**ดับเบิลคลิกที่ cell นี้เพื่อพิมพ์**

- คำตอบ:
- ตัดตัวเลือกไหนทิ้ง เพราะอะไร:

<details>
<summary>✅ เฉลยข้อ 20 (เปิดเมื่อครูบอก)</summary>

**คำตอบ: D. 3**

1. `a > 0` → `5 > 0` จริง → เข้า if ใน
2. `b < 0` → `5 < 0` เท็จ
3. `a > 5` → `5 > 5` **เท็จ** (เท่ากัน ไม่ได้มากกว่า)
4. ไป `else` → `x = 0 + 3 = 3`
- B ✘ คิดว่า `5 > 5` จริง
- C ✘ เข้า `else` ตัวนอก แต่ `a > 0` จริง

</details>

### 📚 ความรู้ที่ต้องรู้

- `if` / `elif` / `else` ทำงาน**แค่ทางเดียว**: ทางแรกที่เงื่อนไขจริง
- การย่อหน้าบอกว่า `else` ตัวไหนคู่กับ `if` ตัวไหน
- ⚠️ `>` กับ `>=` ต่างกัน: `5 > 5` เป็นเท็จ, `5 >= 5` เป็นจริง
- วงเล็บ `if(a > 0):` ใส่ได้ ไม่มีผล
