## ข้อ 3 — ⭐ lambda + ฟังก์ชันคืนค่าฟังก์ชัน + scope

⏱️ เวลาที่แนะนำ: **8 นาที**

What is the output of the program?

````python

a = 2
def func(b):
    b = a * b
    return lambda a: a + b

print(func(4)(2))
````

- A. 0
- B. 244
- C. 422
- D. 10
- E. None of the above

### 📝 คำตอบของฉัน (ห้ามรันโค้ดก่อนตอบ!)
**ดับเบิลคลิกที่ cell นี้เพื่อพิมพ์**

- คำตอบ: 
- ตัดตัวเลือกไหนทิ้ง เพราะอะไร:

### 📚 ความรู้ที่ต้องรู้

**1) `lambda` คือฟังก์ชันสั้น ๆ แบบไม่มีชื่อ**

```python
lambda a: a + b
# เหมือนกับ
def no_name(a):
    return a + b
```

**2) ฟังก์ชันคืนค่าเป็นฟังก์ชันได้**: `func(4)(2)` อ่านเป็น 2 จังหวะ
1. `func(4)` → ได้ **ฟังก์ชัน** ออกมา
2. เอาฟังก์ชันนั้นไปเรียกด้วย `(2)`

**3) ตัวแปรชื่อเดียวกันอาจเป็นคนละตัว (scope)**
- ใน `func`: `a` หมายถึง `a = 2` ข้างนอก (global)
- ใน `lambda a:`: `a` คือ **พารามิเตอร์ของ lambda** ซึ่งบังตัว `a` ข้างนอก
- `b` ใน lambda คือ `b` ของ `func` ที่ lambda **จำไว้** (เรียกว่า **closure**)

**4) ⚠️ กับดัก:** ตัวเลือก 244/422 คือการเอาเลขมาต่อกันเหมือน string แต่ `+` ระหว่างตัวเลขคือการบวกเลข

## ข้อ 4 — วนลูป dict

⏱️ เวลาที่แนะนำ: **3 นาที**

Which of the following options could be filled in line 4 to print the key and value of the dictionary line-by-line?

````python

a = { 1: 2, 3: 4 }

for k, v in a.items():   # ← ลองเปลี่ยนตรงนี้เป็นตัวเลือก A/B/C
    print(k, v)
````

- A. `a:`
- B. `iterate(a):`
- C. `a.items():`
- D. All of the above
- E. None of the above

### 📝 คำตอบของฉัน (ห้ามรันโค้ดก่อนตอบ!)
**ดับเบิลคลิกที่ cell นี้เพื่อพิมพ์**

- คำตอบ: 
- ตัดตัวเลือกไหนทิ้ง เพราะอะไร:

### 📚 ความรู้ที่ต้องรู้

| เขียนว่า | ได้อะไรในแต่ละรอบ |
|---|---|
| `for k in a` | **key** อย่างเดียว |
| `for v in a.values()` | value อย่างเดียว |
| `for k, v in a.items()` | คู่ `(key, value)` แล้ว **แตก (unpack)** ออกเป็น `k`, `v` |

- `k, v = (1, 2)` → `k = 1`, `v = 2` (unpack ได้เพราะฝั่งขวามี 2 ตัว)
- Python **ไม่มี** ฟังก์ชันชื่อ `iterate()`

## ข้อ 5 — ⭐ Recursion (ฟังก์ชันเรียกตัวเอง)

⏱️ เวลาที่แนะนำ: **7 นาที**

Following is a definition of a recursive function, what is the value of `recur(-14)`?

````python
def recur(i):
    if i > 0:
        return recur(i - 1)
    if i < 0:
        return recur(5)
    return i

print(recur(-14))
````

- A. The recursive function never terminates
- B. -1
- C. 5
- D. 0
- E. None of the above

### 📝 คำตอบของฉัน (ห้ามรันโค้ดก่อนตอบ!)
**ดับเบิลคลิกที่ cell นี้เพื่อพิมพ์**

- คำตอบ: 
- ตัดตัวเลือกไหนทิ้ง เพราะอะไร:

### 📚 ความรู้ที่ต้องรู้

**Recursion** = ฟังก์ชันที่เรียกตัวเอง ต้องมี 2 ส่วนเสมอ
1. **Base case**: จุดหยุด ไม่เรียกตัวเองอีก (ในข้อนี้คือ `return i` ตอน `i == 0`)
2. **Recursive case**: เรียกตัวเองด้วยค่าที่ **เข้าใกล้ base case** มากขึ้น

**วิธีทำในห้องสอบ: เขียนเส้นทางการเรียกเป็นลูกศร**
```
recur(-14) → recur(5) → recur(4) → recur(3) → recur(2) → recur(1) → recur(0) = 0
```
ค่าที่ `return` จะถูกส่งย้อนกลับขึ้นไปจนถึงตัวแรก

⚠️ ถ้าไม่มีทางถึง base case → `RecursionError` (เรียกตัวเองลึกเกินประมาณ 1000 ชั้น)

### 🔍 ดูการทำงานทีละบรรทัดใน Python Tutor

👉 [เปิดโค้ดข้อนี้ใน Python Tutor](https://pythontutor.com/visualize.html#code=def%20recur(i)%3A%0A%20%20%20%20if%20i%20%3E%200%3A%0A%20%20%20%20%20%20%20%20return%20recur(i%20-%201)%0A%20%20%20%20if%20i%20%3C%200%3A%0A%20%20%20%20%20%20%20%20return%20recur(5)%0A%20%20%20%20return%20i%0A%0Aprint(recur(-14))&cumulative=false&curInstr=0&heapPrimitives=nevernest&mode=display&origin=opt-frontend.js&py=3&rawInputLstJSON=%5B%5D&textReferences=false)

ดู **frame** ซ้อนกันลงไปเรื่อย ๆ ทางขวา (call stack) แล้วสังเกตตอนที่ค่า `0` ถูกส่งกลับขึ้นไปทีละชั้น

### 🧪 ทดลอง: ทายก่อน แล้วค่อยรัน

**ทดลอง 5.1** เพิ่ม `print` เพื่อดูเส้นทางการเรียก จะพิมพ์ออกมากี่บรรทัด?

✍️ ฉันทายว่า:

````python
def recur_trace(i):
    print(f"recur({i})")
    if i > 0:
        return recur_trace(i - 1)
    if i < 0:
        return recur_trace(5)
    return i

print("ผลลัพธ์ =", recur_trace(-14))
````

## ข้อ 15 — ⭐ กับดัก mutable default argument

⏱️ เวลาที่แนะนำ: **7 นาที**

What is the value of `arr` after executing this code?

````python

def func(a = [1,2,3]):
    a[1] = 0
    return a

arr = func()
arr[1] = 3
arr[2] = 5
arr = func()
print(arr)
````

- A. 0
- B. [1, 2, 3]
- C. [1, 3, 5]
- D. [1, 0, 5]
- E. None of the above

### 📝 คำตอบของฉัน (ห้ามรันโค้ดก่อนตอบ!)
**ดับเบิลคลิกที่ cell นี้เพื่อพิมพ์**

- คำตอบ: 
- ตัดตัวเลือกไหนทิ้ง เพราะอะไร:

### 📚 ความรู้ที่ต้องรู้

**1) ค่า default ถูกสร้างแค่ "ครั้งเดียว" ตอนที่ `def` ทำงาน** ไม่ได้สร้างใหม่ทุกครั้งที่เรียกฟังก์ชัน

**2) list แก้ไขได้ (mutable)** → ทุกครั้งที่เรียก `func()` โดยไม่ส่งค่า จะได้ **list ก้อนเดิม** ที่ใช้ร่วมกันทุกครั้ง

**3) `arr = func()`** ไม่ได้คัดลอก list แต่ทำให้ `arr` **ชี้ไปที่ list ก้อนเดียวกัน** การแก้ `arr[1]` จึงเท่ากับแก้ค่า default ไปด้วย

**4) วิธีแก้มาตรฐาน**: ใช้ `None` เป็นค่า default
```python
def func(a=None):
    if a is None:
        a = [1, 2, 3]   # สร้างใหม่ทุกครั้งที่เรียก
```

## 1) ควรเดาไหม? (ถูก +4 · ผิด −1 · ไม่ตอบ 0)

````python
for k in range(4):
    n = 5 - k                      # ตัวเลือกที่เหลือ
    ev = 4 * (1/n) - 1 * ((n-1)/n)  # คะแนนคาดหวัง
    print(f"ตัดทิ้งได้ {k} ตัว → เหลือ {n} ตัว → คะแนนคาดหวัง {ev:+.2f}")
````

📌 **กฎ:**
- **ตัดทิ้งได้อย่างน้อย 1 ตัว → เดาเลย** (คะแนนคาดหวังเป็นบวก)
- **ตัดทิ้งไม่ได้เลย → เว้นไว้** (เดาก็ได้คะแนนคาดหวัง 0 เท่ากับไม่ตอบ แต่ผลเสี่ยงกว่า)
