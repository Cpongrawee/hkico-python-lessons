### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
**dict** (dictionary) เก็บข้อมูลเป็นคู่ **key → value** เหมือนสมุดโทรศัพท์: รู้ชื่อ (key) ก็เปิดหาเบอร์ (value) ได้ทันที ไม่ต้องไล่หาทีละช่องแบบ list

#### 2) กฎที่ต้องจำ
- รูปแบบ: `{key1: value1, key2: value2}` → `:` คั่น key กับ value, `,` คั่นแต่ละคู่
- key ที่เป็นข้อความต้องมี `'...'` หรือ `"..."` ถ้าไม่มี Python จะไปหา **ตัวแปร** ชื่อนั้น
- key ต้องเป็นค่าที่แก้ไขไม่ได้ เช่น `str`, `int`, `tuple` (ใช้ list เป็น key ไม่ได้)
- key ห้ามซ้ำ: ถ้าเขียนซ้ำ ตัวหลังจะ**ทับ**ตัวหน้า
- อ่านค่า `abc['a']` · เพิ่ม/แก้ `abc['d'] = 78`
- อีกวิธีสร้าง: `dict(a=12, b=34)` (แบบนี้ใช้ `=` ได้ เพราะเป็นการส่ง argument ให้ฟังก์ชัน)

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
```python
abc = {'a': 12, 'b': 34}
abc['c'] = 56
abc['a'] = 99
print(abc)
```

| บรรทัด | `abc` หลังทำบรรทัดนี้ |
|---|---|
| 1 | `{'a': 12, 'b': 34}` |
| 2 | `{'a': 12, 'b': 34, 'c': 56}` (key ใหม่ → เพิ่มต่อท้าย) |
| 3 | `{'a': 99, 'b': 34, 'c': 56}` (key เดิม → แก้ค่า) |
| 4 | พิมพ์ `{'a': 99, 'b': 34, 'c': 56}` |

#### 4) กับดักที่เจอบ่อย
- สับสน `=` กับ `:` (ตัวเลือก A): ใน `{ }` ต้องเป็น `:` เสมอ
- ลืมเครื่องหมายคำพูด (ตัวเลือก C): `{a: 12}` ไม่ใช่ key ชื่อ `'a'` แต่เป็น "ค่าของตัวแปร `a`"
- เห็นว่ามีตัวเลือกที่ถูก 1 ข้อแล้วกด D ทันที: "All the above" ต้องถูก**ทุก**ตัวเลือก

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
เหมือนคณิตศาสตร์ที่เรียน: **วงเล็บก่อน → ยกกำลัง → คูณหาร → บวกลบ** Python มีเครื่องหมายมากกว่าคณิต (`//`, `%`, `and`, `or`) จึงต้องรู้ว่าแต่ละตัวอยู่ชั้นไหน

#### 2) กฎที่ต้องจำ
| ลำดับ (บนทำก่อน) | เครื่องหมาย |
|---|---|
| 1 | `( )` |
| 2 | `**` (ยกกำลัง: ทำจาก**ขวาไปซ้าย**) |
| 3 | `+x`, `-x` (เครื่องหมายบวก/ลบหน้าตัวเลข) |
| 4 | `*`, `/`, `//`, `%` (ระดับเดียวกัน ทำจากซ้ายไปขวา) |
| 5 | `+`, `-` |
| 6 | `==`, `!=`, `<`, `>`, `<=`, `>=` |
| 7 | `not` → `and` → `or` |

⚠️ `-2 ** 2` = `-(2 ** 2)` = `-4` เพราะ `**` อยู่สูงกว่าเครื่องหมายลบหน้าตัวเลข

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
```python
print(2 + 3 * 4 ** 2 % 5)
```

| ขั้น | ทำอะไร | ได้ |
|---|---|---|
| 1 | `4 ** 2` (ยกกำลังก่อน) | `2 + 3 * 16 % 5` |
| 2 | `3 * 16` (`*` กับ `%` ระดับเดียวกัน → ซ้ายก่อน) | `2 + 48 % 5` |
| 3 | `48 % 5` | `2 + 3` |
| 4 | `2 + 3` | `5` |

#### 4) กับดักที่เจอบ่อย
- คิดว่าคูณมาก่อนยกกำลัง (ตัวเลือก B)
- คิดว่า `%` อยู่ระดับต่ำกว่า `*`: ที่จริงอยู่ระดับเดียวกัน
- `2 ** 3 ** 2` ทำจากขวา: `2 ** 9 = 512` ไม่ใช่ `8 ** 2 = 64`

**คำตอบ: A. The code doesn't have errors** (พิมพ์ `23.1`)

1. `+x +y` → `+x` คือ `20` แล้ว `20 + 3.1` → `23.1` (บวกทำก่อน `and`)
2. `"sum =" and 23.1` → ข้อความไม่ว่างนับเป็นจริง → `and` คืน**ตัวขวา** → `23.1`

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
`and` / `or` ใน Python **ไม่ได้คืน True/False เสมอไป** มันคืน "ค่าตัวใดตัวหนึ่ง" ที่อยู่ข้าง ๆ มัน
ส่วน `;` เป็นแค่ตัวคั่นคำสั่ง (เหมือนขึ้นบรรทัดใหม่)

#### 2) กฎที่ต้องจำ
- **falsy** (นับเป็นเท็จ): `False`, `0`, `0.0`, `''`, `[]`, `{}`, `None` · ค่าอื่นทั้งหมดเป็น **truthy**
- `a and b`: ถ้า `a` เป็นเท็จ → คืน `a` · ถ้า `a` เป็นจริง → คืน `b`
- `a or b`: ถ้า `a` เป็นจริง → คืน `a` · ถ้า `a` เป็นเท็จ → คืน `b`
- `+x` = `x` · `-x` = ติดลบ (เครื่องหมายหน้าตัวเลข)
- `and` อยู่ลำดับต่ำกว่า `+` จึงบวกให้เสร็จก่อน: `"sum =" and +x +y` = `"sum =" and (x + y)`
- `x=20;` → `;` คั่นคำสั่งได้ เช่น `a = 1; b = 2` ในบรรทัดเดียว

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด

|---|---|---|

| 2 | `20 + y` | `23.1` |
| 3 | `"sum ="` เป็นจริงไหม? | จริง (ข้อความไม่ว่าง) |
| 4 | `and` คืนตัวขวา | `23.1` |
| 5 | `print(23.1)` | พิมพ์ `23.1` |

#### 4) กับดักที่เจอบ่อย
- คิดว่า `and` คืน `True` เลยตอบว่าพิมพ์ `True`
- คิดว่าพิมพ์ `sum = 23.1`: ถ้าอยากได้แบบนั้นต้องเขียน `print("sum =", x + y)`
- เห็น `;` แล้วตกใจว่า error (ตัวเลือก B): Python ยอมรับ

## ข้อ 6 — ⭐ กับดัก: ลืม `:` + ตัวแปรในฟังก์ชัน

What is the expected output of the following code? *(ดูทุกตัวอักษรให้ดี)*

````python
x=3;
y=7%5;

def main()
    x=5
    y=2

main()

if y==2;
    x=3
print(x+y)
````

- A. 4
- B. 5
- C. 6
- D. 7
- E. None of the above

### 📝 คำตอบของฉัน (ห้ามรันโค้ดก่อนตอบ!)
**ดับเบิลคลิกที่ cell นี้เพื่อพิมพ์**

- คำตอบ:
- ตัดตัวเลือกไหนทิ้ง เพราะอะไร:

<details>
<summary>✅ เฉลยข้อ 6 (เปิดเมื่อครูบอก)</summary>

**คำตอบ: E. None of the above** (โค้ดนี้ `SyntaxError` ไม่มี output เลย)

นี่คือ**กับดัก**: `def main()` ไม่มี `:` และ `if y==2;` ใช้ `;` แทน `:` → Python ไม่ยอมรันแม้แต่บรรทัดแรก

> ถ้าเป็นแค่พิมพ์ผิด (มี `:` ครบ) จะได้ **B. 5**: `y = 7 % 5 = 2` · `main()` เปลี่ยนแค่ตัวแปร**ข้างในฟังก์ชัน** · `y==2` จริง → `x = 3` → `3 + 2 = 5`

- D ✘ (7) คิดว่า `main()` เปลี่ยน `x` ข้างนอกเป็น 5 → `5 + 2`
- A, C ✘ คำนวณ `7 % 5` ผิด

</details>

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
ข้อนี้ทดสอบ 2 เรื่อง:
1. **อ่านทุกตัวอักษร**: ข้อสอบจงใจวางกับดักเล็ก ๆ เช่น ลืม `:` ถ้าอ่านเร็วจะตอบ 5 ทันที
2. **ตัวแปรในฟังก์ชันเป็นของฟังก์ชันเท่านั้น** (local) ไม่ไปเปลี่ยนตัวแปรชื่อเดียวกันข้างนอก

#### 2) กฎที่ต้องจำ
- `def ...`, `if ...`, `elif ...`, `else`, `for ...`, `while ...`, `class ...` ต้องจบด้วย `:` เสมอ
- **SyntaxError ถูกตรวจทั้งไฟล์ก่อนเริ่มรัน** → ถ้ามีจุดเดียวผิด จะไม่มีบรรทัดไหนทำงานเลย ไม่มี output แม้แต่บรรทัดแรก
- `;` ท้ายบรรทัดใช้ได้ แต่ใช้แทน `:` ไม่ได้
- ตัวแปรที่**กำหนดค่า**ในฟังก์ชัน (`x=5`) เป็นตัวแปรใหม่ของฟังก์ชัน หายไปเมื่อฟังก์ชันจบ
- ถ้าอยากเปลี่ยนตัวแปรข้างนอกจริง ๆ ต้องเขียน `global x` ในฟังก์ชัน

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
ถ้าแก้ `:` ให้ถูกแล้ว:

| บรรทัด | ทำอะไร | `x` ข้างนอก | `y` ข้างนอก |
|---|---|---|---|
| `x=3;` | | 3 | — |
| `y=7%5;` | 7 หาร 5 เหลือเศษ 2 | 3 | 2 |
| `main()` | สร้าง `x=5`, `y=2` **ของ main** แล้วทิ้งไปเมื่อจบ | 3 | 2 |
| `if y==2:` | `2 == 2` จริง | | |
| `x=3` | | 3 | 2 |
| `print(x+y)` | `3 + 2` | พิมพ์ **5** | |

#### 4) กับดักที่เจอบ่อย
- ไม่เห็นว่าไม่มี `:` → ตอบ B ทันที
- คิดว่า SyntaxError ทำให้ "รันได้ถึงบรรทัดที่ผิดแล้วค่อยหยุด": ผิด ไม่รันเลยสักบรรทัด
- คิดว่า `main()` เปลี่ยน `x` ข้างนอก → ตอบ D (7)

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
ฟังก์ชันคณิตศาสตร์เพิ่มเติมอยู่ใน **module `math`** ต้อง `import math` ก่อน แล้วเรียกแบบ `math.ชื่อฟังก์ชัน(...)`

#### 2) กฎที่ต้องจำ
| ฟังก์ชัน | ทำอะไร | ตัวอย่าง |
|---|---|---|
| `math.sqrt(x)` | รากที่สอง (ได้ float เสมอ) | `math.sqrt(16)` → `4.0` |
| `math.floor(x)` | ปัด**ลง** | `math.floor(2.7)` → `2` |

| `x ** (1/n)` | รากที่ n (ไม่ต้องใช้ math) | `16 ** 0.5` → `4.0` |

⚠️ ตัวเลขติดลบ: `floor` ปัดไปทางลบ, `int()` ตัดทศนิยมทิ้ง (เข้าหา 0)

| `int(-2.5)` | ตัดทศนิยมทิ้ง | `-2` |
| `math.ceil(-2.5)` | จำนวนเต็มที่**มากกว่าหรือเท่ากับ** -2.5 | `-2` |

## ข้อ 8 — ลูปสะสมค่า + `int()` ตัดทศนิยม

What is the output?

````python
sum = 0
count = 0
for value in [9,41,12,3,74,15]:
    sum+=value
    count+=1

print(sum, count, int(sum/count))
````

- A. 154 6 25
- B. 155 6 24
- C. 154 5 24
- D. 155 5 25
- E. None of the above

### 📝 คำตอบของฉัน (ห้ามรันโค้ดก่อนตอบ!)
**ดับเบิลคลิกที่ cell นี้เพื่อพิมพ์**

- คำตอบ:
- ตัดตัวเลือกไหนทิ้ง เพราะอะไร:

<details>
<summary>✅ เฉลยข้อ 8 (เปิดเมื่อครูบอก)</summary>

**คำตอบ: A. 154 6 25**

- `sum` = 9+41+12+3+74+15 = **154**
- `count` = มี 6 ตัว → **6**
- `154 / 6` = 25.666... → `int()` **ตัดทศนิยมทิ้ง** → **25**
- B, D ✘ บวกผิด · C ✘ นับขาด · ถ้าคิดว่า `int()` ปัดเศษ จะได้ 26 (ไม่มีในตัวเลือก)

</details>

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
**รูปแบบสะสมค่า** (accumulator): เริ่มตัวแปรที่ 0 แล้วบวกเพิ่มทุกรอบของลูป ใช้หาผลรวม นับจำนวน และหาค่าเฉลี่ย

#### 2) กฎที่ต้องจำ
- `sum += value` คือ `sum = sum + value`
- `/` ให้ผลเป็น **float เสมอ** (แม้หารลงตัว: `6/2` → `3.0`)
- `int(25.67)` → `25`: **ตัดทิ้ง ไม่ปัด** · ถ้าอยากปัดใช้ `round()`
- `//` หารปัดลง: `154 // 6` → `25`
- ⚠️ ตั้งชื่อตัวแปรว่า `sum` ได้ แต่จะ**บัง**ฟังก์ชัน `sum()` ที่มีอยู่แล้ว ใช้ `sum([...])` ต่อไม่ได้

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
| รอบ | `value` | `sum` หลังรอบ | `count` หลังรอบ |
|---|---|---|---|
| 1 | 9 | 9 | 1 |
| 2 | 41 | 50 | 2 |
| 3 | 12 | 62 | 3 |
| 4 | 3 | 65 | 4 |
| 5 | 74 | 139 | 5 |
| 6 | 15 | 154 | 6 |

`int(154 / 6)` = `int(25.666...)` = **25**

#### 4) กับดักที่เจอบ่อย
- บวกในหัวผิด: จดค่าสะสมทีละรอบแบบตารางข้างบน
- คิดว่า `int()` ปัดเศษขึ้น
- ลืมว่า `count += 1` อยู่ในลูป (ย่อหน้าเดียวกับ `sum`) จึงนับครบทุกรอบ

## ข้อ 10 — ต่อ string แล้วเลือกตัวด้วย index

Given the code below, which of the following is code correct if the output is `I`?

````python
x='HKICO'
y='2022'

var=x+y
````

- A. print(x[3])
- B. print(x[4])
- C. print(var[2])
- D. print(var[3])
- E. None of the above

### 📝 คำตอบของฉัน (ห้ามรันโค้ดก่อนตอบ!)
**ดับเบิลคลิกที่ cell นี้เพื่อพิมพ์**

- คำตอบ:
- ตัดตัวเลือกไหนทิ้ง เพราะอะไร:

<details>
<summary>✅ เฉลยข้อ 10 (เปิดเมื่อครูบอก)</summary>

**คำตอบ: C. print(var[2])**

`var = 'HKICO2022'` · index เริ่มที่ 0: `H`=0, `K`=1, **`I`=2**, `C`=3, `O`=4
- A ✘ `x[3]` = `'C'` (นับจาก 1 จะได้ I ซึ่งผิด) · B ✘ `x[4]` = `'O'` · D ✘ `var[3]` = `'C'`

</details>

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
`+` ระหว่าง string คือ**ต่อกัน** และตำแหน่งตัวอักษรใน string (index) **เริ่มนับที่ 0**

#### 2) กฎที่ต้องจำ
- `'HKICO' + '2022'` → `'HKICO2022'` (`'2022'` มีเครื่องหมายคำพูด จึงเป็น string ไม่ใช่ตัวเลข)
- index: ตัวแรก = 0, ตัวสุดท้าย = `len - 1`
- index ติดลบนับจากท้าย: `var[-1]` = ตัวสุดท้าย
- index เกินความยาว → `IndexError`
- string + int → `TypeError` ต้องแปลงก่อน: `'HKICO' + str(2022)`

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
| index | 0 | 1 | 2 | 3 | 4 | 5 | 6 | 7 | 8 |

#### 2) กฎที่ต้องจำ
- `s.split()` (ไม่ใส่อะไร) แยกตามช่องว่างทุกแบบ ช่องว่างติดกันหลายช่องนับเป็นหนึ่ง

## ข้อ 12–14 — ⭐ recursion (factorial) + `int(input())` (ใช้โค้ดเดียวกัน)

````python
x=int(input('Input a number:'))

def RE(x):
    if(x==1):
        result = x
    else:
        result = x * RE(x-1)
    return result

print(RE(x))
````

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
**recursion** = ฟังก์ชันที่เรียกตัวเอง โดยแต่ละครั้งทำโจทย์ให้เล็กลง จนถึง**กรณีฐาน** (base case) ที่ตอบได้ทันทีโดยไม่เรียกตัวเองอีก
`RE(x)` คือ factorial: `RE(5) = 5 × RE(4)` และ `RE(1) = 1`

อีกเรื่องที่ข้อนี้ทดสอบคือ `input()` ซึ่ง**คืนค่าเป็น string เสมอ**

#### 2) กฎที่ต้องจำ
- `input()` คืน string เสมอ แม้ผู้ใช้พิมพ์ตัวเลข
- `int('5')` → `5` · `int(' 7 ')` → `7` · `int('-3')` → `-3`
- `int('1.1')`, `int('5.0')`, `int('x')` → **`ValueError`** (ถ้าจะแปลงทศนิยม ต้อง `int(float('1.1'))` → `1`)
- error ที่บรรทัดแรก = โปรแกรมหยุดทันที บรรทัดต่อไปไม่ทำงาน
- recursion ต้องมี 2 ส่วน: **กรณีฐาน** (`x==1` → คืน `x`) และ**ขั้นที่ทำให้เล็กลง** (`x * RE(x-1)`)
- ถ้าไม่มีวันถึงกรณีฐาน → เรียกตัวเองไปเรื่อย ๆ จน `RecursionError`

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
`RE(5)`: แต่ละครั้ง**รอ**คำตอบจากครั้งที่เล็กกว่า แล้วค่อยคูณตอนขากลับ

| การเรียก | รอคำตอบจาก | คืนค่า (ขากลับ ล่างขึ้นบน) |
|---|---|---|
| `RE(5)` | `5 * RE(4)` | 5 × 24 = **120** |
| `RE(4)` | `4 * RE(3)` | 4 × 6 = 24 |
| `RE(3)` | `3 * RE(2)` | 3 × 2 = 6 |
| `RE(2)` | `2 * RE(1)` | 2 × 1 = 2 |
| `RE(1)` | กรณีฐาน | 1 |

#### 4) กับดักที่เจอบ่อย
- ลืมว่า `int(input())` พังได้ (ข้อ 12, 13): ดูข้อมูลที่ป้อนก่อนคิดเรื่อง recursion
- คิดว่า `int('1.1')` ได้ 1: ต้องผ่าน `float` ก่อน
- คำนวณ factorial แค่สองขั้น (ข้อ 14 ตัวเลือก B)

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
**class** คือแม่แบบ ส่วน **object** คือของที่สร้างจากแม่แบบนั้น
- `__init__` ถูกเรียก**อัตโนมัติ**ตอนสร้าง object (`Add(1000, 2000)`)
- method อื่น (`cal`, `dis`) ทำงาน**เฉพาะตอนที่เราเรียก**เท่านั้น

#### 2) กฎที่ต้องจำ
- ตัวแปรที่เขียนใต้ `class` ตรง ๆ (`ans=0`) = **class attribute** เป็นค่าเริ่มต้นที่ทุก object ใช้ร่วมกัน
- `self.xxx = ...` ใน method = **instance attribute** ของ object นั้นเท่านั้น
- อ่าน `self.ans`: หาใน object ก่อน ถ้าไม่มีไปหาที่ class
- `Add(1000, 2000)` ส่ง `1000, 2000` ไปให้ `__init__(self, f, s)` (`self` คือ object ใหม่ ส่งให้เอง)
- ชื่อต้องเป็น `__init__` (ขีดล่างสองตัวหน้าและหลัง) ถ้าชื่ออื่น Python ไม่เรียกให้
- class ที่ไม่มี `__init__` ของตัวเองรับ argument ไม่ได้ → `TypeError: ... takes no arguments`

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
| ขั้น | ทำอะไร | `obj.fir` | `obj.sen` | `obj.ans` |
|---|---|---|---|---|
| `obj=Add(1000,2000)` | เรียก `__init__` อัตโนมัติ | 1000 | 2000 | 0 (จาก class) |
| `obj.dis()` | พิมพ์ `self.ans` | | | พิมพ์ **0** |
| *(ถ้าเรียก `obj.cal()` ก่อน)* | `self.ans = 1000 + 2000` | 1000 | 2000 | 3000 |

#### 4) กับดักที่เจอบ่อย
- คิดว่าทุก method ทำงานเองตอนสร้าง object (ข้อ 15 ตัวเลือก C)
- คิดว่าเปลี่ยนชื่อ `__init__` แล้วแค่ "ไม่ได้ตั้งค่า" แต่ไม่ error: error เพราะส่ง argument ไปแล้วไม่มีใครรับ
- สับสนระหว่าง SyntaxError (เขียนผิดไวยากรณ์ รันไม่ได้เลย) กับ error ตอนรัน (เช่น TypeError)

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
class สามารถกำหนดได้เองว่า
- `print(obj)` จะแสดงอะไร → เขียนใน `__str__`
- `obj1 + obj2` จะได้อะไร → เขียนใน `__add__`

ส่วน `'...%d...' % (ค่า)` คือการแทนค่าลงในข้อความ (string formatting แบบเก่า)

#### 2) กฎที่ต้องจำ
- `v1 + v2` → Python เรียก `v1.__add__(v2)` (`self` = `v1`, `other` = `v2`)
- `print(obj)` → Python เรียก `obj.__str__()` แล้วพิมพ์ข้อความที่ได้
- ไม่มี `__str__` → พิมพ์แบบ `<__main__.Hello object at 0x...>`

| รหัส | แทนด้วย | `'%?' % 65` |
|---|---|---|
| `%d` | จำนวนเต็ม (float ถูกตัดทศนิยม) | `'65'` |
| `%s` | ข้อความ (`str(...)`) | `'65'` |
| `%f` | ทศนิยม 6 ตำแหน่ง | `'65.000000'` |
| `%c` | ตัวอักษรของรหัสนั้น (`chr`) | `'A'` |

ค่าหลาย ๆ ตัวใส่เป็น tuple เรียงตามลำดับ `%`

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
| ขั้น | ทำอะไร | ได้ |
|---|---|---|
| 1 | `v1 = Hello(2, 10)` | `v1.a=2`, `v1.b=10` |
| 2 | `v2 = Hello(5, -2)` | `v2.a=5`, `v2.b=-2` |
| 3 | `v1 + v2` → `v1.__add__(v2)` | `Hello(2+5, 10+(-2))` = `Hello(7, 8)` |
| 4 | `print(...)` → `__str__` | `'Hello (%d, %d)' % (7, 8)` |
| 5 | แทนค่า | `Hello (7, 8)` |

#### 4) กับดักที่เจอบ่อย
- คิดว่า `+` ระหว่าง object ทำไม่ได้ (ตัวเลือก D ของข้อ 17): ทำได้เพราะมี `__add__`
- คิดว่า `print` พิมพ์ทั้ง `v1` และ `v2` (ตัวเลือก A)
- `%c` ของตัวเลขเล็ก ๆ (0–31) ได้ตัวควบคุมที่มองไม่เห็น ไม่ใช่ตัวเลข

#### 🧠 เช็คตัวเอง
ทายผลของทั้งสามช่อง

```python
print('%d|%s|%c' % (3.9, 3.9, 65))
```

<summary>เฉลยเช็คตัวเอง</summary>

```

```

`%d` ตัด 3.9 เหลือ 3 · `%s` แสดงตามเดิม · `%c` ของ 65 คือ `A`

</details>

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
**Fibonacci**: แต่ละตัวคือผลบวกของสองตัวก่อนหน้า `0, 1, 1, 2, 3, 5, 8, ...`
ฟังก์ชันนี้เรียกตัวเอง**สองครั้ง**ต่อรอบ และมีกรณีฐานสองกรณี (0 และ 1)

#### 2) กฎที่ต้องจำ
- `if` → ต้องมีเงื่อนไข · `elif` → ต้องมีเงื่อนไข · `else` → **ห้าม**มีเงื่อนไข
- ทำงานแค่ทางแรกที่เงื่อนไขจริง
- `range(0, n)` = 0, 1, ..., n-1 (**ไม่รวม n**)
- `range(0, ตัวเลขที่ ≤ 0)` = ว่าง → ลูปไม่ทำงาน (ไม่ error)
- `int('5.0')` → `ValueError` (เหมือนข้อ 13)
- ย่อหน้ากี่ช่องก็ได้ (ในข้อนี้ใช้ 8 ช่อง) ขอแค่ในบล็อกเดียวกันเท่ากัน

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
ถ้าป้อน `5`: `for n in range(0, 5)` → n = 0, 1, 2, 3, 4

| `n` | `main(n)` คิดยังไง | พิมพ์ |
|---|---|---|
| 0 | กรณีฐาน | 0 |
| 1 | กรณีฐาน | 1 |
| 2 | `main(0) + main(1)` = 0 + 1 | 1 |
| 3 | `main(1) + main(2)` = 1 + 1 | 2 |
| 4 | `main(2) + main(3)` = 1 + 2 | 3 |

รวม 0 + 1 + 1 + 2 + 3 = 7

#### 4) กับดักที่เจอบ่อย
- ไม่เห็นว่า `5.0` ผ่าน `int()` ไม่ได้ (ข้อ 19)
- นับ `range(0, 5)` รวม 5 → ได้ `0 1 1 2 3 5` รวม 12
- คิดว่าเลขติดลบทำให้ recursion ไม่จบ: ลูปไม่ได้เรียก `main` เลยสักครั้ง
