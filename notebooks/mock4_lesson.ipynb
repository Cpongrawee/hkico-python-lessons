`%.2f` = แสดงเป็นทศนิยม **2 ตำแหน่ง** โดย**ปัดเศษ**: ตำแหน่งที่ 3 คือ 6 (≥ 5) จึงปัดขึ้น → `56.24`
- A ✘ ไม่ได้ตัดตำแหน่งเลย · B ✘ ตัดทิ้ง ไม่ได้ปัด · C ✘ คือรูปแบบ `%.4f` ของเลขอื่น

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
`"%.2f" % x` คือการ**จัดรูปแบบตัวเลข**ให้เป็นข้อความ: เอา `x` ไปใส่แทนที่ `%.2f` โดยแสดงทศนิยมตามจำนวนที่กำหนด ใช้บ่อยเวลาแสดงราคาหรือคะแนน

#### 2) กฎที่ต้องจำ
- `%f` = ทศนิยม 6 ตำแหน่ง (ค่าเริ่มต้น) · `%.2f` = 2 ตำแหน่ง · `%.0f` = ไม่มีทศนิยม
- **ปัดเศษ** ไม่ใช่ตัดทิ้ง: ดูตำแหน่งถัดไป ถ้า ≥ 5 ปัดขึ้น
- ผลที่ได้เป็น **string** ไม่ใช่ตัวเลข
- ใส่ตัวเลขจำนวนเต็มก็ได้: `"%.2f" % 5` → `'5.00'`
- ⚠️ float เก็บค่าไม่ตรงเป๊ะ บางค่าเช่น 2.675 จึงปัดได้ 2.67 (ค่าจริงในเครื่องคือ 2.67499999...)

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
```python
x = 56.236
print("%.2f" % x)
```

| ขั้น | ทำอะไร | ได้ |
|---|---|---|
| 1 | ดูทศนิยม 2 ตำแหน่งแรก | `56.23` |
| 2 | ดูตำแหน่งที่ 3 | `6` |
| 3 | `6 ≥ 5` → ปัดขึ้น | `56.24` |
| 4 | ได้ string แล้ว `print` | พิมพ์ `56.24` |

#### 4) กับดักที่เจอบ่อย
- คิดว่า `%.2f` ตัดทิ้งเฉย ๆ (ตัวเลือก B)
- สับสนว่าตัวเลขหลัง `.` คือจำนวนตำแหน่งทศนิยม ไม่ใช่จำนวนหลักทั้งหมด
- ลืมว่าผลเป็น string: `"%.2f" % x + 1` จะ error

#### 🧠 เช็คตัวเอง
ทายผลทั้งสามค่า

```python
print("%.1f" % 3.96, "%.0f" % 7.6, "%.3f" % 2)
```

</details>

## ข้อ 2 — ต่อ string ด้วย `+`

What will be the output of the following Python statement?

> ⚠️ ข้อสอบต้นฉบับ**ไม่ได้พิมพ์โค้ดของข้อนี้มา** (มีแต่ตัวเลือก) โค้ดข้างล่างคือโค้ดที่เข้ากับตัวเลือก A–D ทั้งหมด

````python
print("a" + "bc")
````

- A. a
- B. bc
- C. abc
- D. bca
- E. None of the above

### 📝 คำตอบของฉัน (ห้ามรันโค้ดก่อนตอบ!)

<details>

**คำตอบ: C. abc**

`+` ระหว่าง string คือ**ต่อกัน**ตามลำดับซ้ายไปขวา → `"a"` ตามด้วย `"bc"` = `"abc"`
- D ✘ ต่อกลับด้าน (`"bc" + "a"`) · A, B ✘ ได้แค่ชิ้นเดียว

</details>

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
`+` ทำงานต่างกันตามชนิดข้อมูล: ตัวเลขคือ**บวก** แต่ string คือ**ต่อกัน** (concatenation)

#### 2) กฎที่ต้องจำ
- `"a" + "bc"` → `"abc"` (ตัวซ้ายมาก่อนเสมอ)
- ไม่มีช่องว่างเพิ่มให้: `"Hi" + "Ann"` → `"HiAnn"`
- `*` กับ string คือทำซ้ำ: `"ab" * 3` → `"ababab"` · `"ab" * 0` → `""`
- string + ตัวเลข → `TypeError` ต้องแปลงก่อน `"a" + str(1)`

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
```python
s = "a" + "bc"
s = s + "!"
print(s)
```

| บรรทัด | `s` |
|---|---|
| 1 | `"abc"` |
| 2 | `"abc!"` |
| 3 | พิมพ์ `abc!` |

#### 4) กับดักที่เจอบ่อย
- ต่อกลับด้าน (ตัวเลือก D)
- คิดว่ามีช่องว่างคั่นให้อัตโนมัติ (แบบ `print("a", "bc")` ที่คั่นด้วยช่องว่าง)
- ลืมว่า `"1" + "2"` ได้ `"12"` ไม่ใช่ `3`

#### 🧠 เช็คตัวเอง
ทายผล

```python
print("1" + "2", 1 + 2, "ab" * 2 + "c")
```

<details>
<summary>เฉลยเช็คตัวเอง</summary>

```
12 3 ababc
```

`"1" + "2"` ต่อกันได้ `12` · `1 + 2` บวกเลขได้ `3` · `*` ทำก่อน `+` ได้ `ababc`

</details>

**คำตอบ: B. `//`**

`//` คือหารแล้วทิ้งเศษ ได้จำนวนเต็ม (`7 // 2` → `3`)
- C ✘ `/` หารแบบมีทศนิยม (`7 / 2` → `3.5`) · D ✘ `%` คือ**เศษ**ของการหาร · A ✘ `|` คือ bitwise OR

> ข้อสังเกต: ชื่อที่ถูกต้องตามเอกสาร Python คือ **floor division** (หารปัดลง) ซึ่งต่างจาก "ตัดทิ้ง" ตอนเป็นเลขติดลบ (ดูทดลอง 3.5)

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
การหารใน Python มี 3 แบบ ที่ต้องแยกให้ออก: `/` ได้ทศนิยม, `//` ได้จำนวนเต็มที่**ปัดลง**, `%` ได้**เศษ**

#### 2) กฎที่ต้องจำ
| เครื่องหมาย | ชื่อ | `7 ? 2` | `-7 ? 2` |
|---|---|---|---|
| `/` | หารปกติ (ได้ float เสมอ) | `3.5` | `-3.5` |
| `//` | หารปัดลง (floor division) | `3` | `-4` |
| `%` | เศษ | `1` | `1` |

- `//` ปัด**ลง**ไปทางลบ ไม่ใช่ตัดทิ้งเข้าหา 0
- ถ้ามี float อยู่ในการหาร `//` ได้ float: `7.5 // 2` → `3.0`
- `(a // b) * b + (a % b) == a` เสมอ

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
```python

print(a / b, a // b, a % b)
```

| นิพจน์ | คิดยังไง | ได้ |
|---|---|---|
| `17 / 5` | หารปกติ | `3.4` |
| `17 // 5` | 5 ใส่ใน 17 ได้กี่ครั้งเต็ม ๆ | `3` |
| `17 % 5` | เหลือเศษเท่าไร | `2` |
| ตรวจ | `3 * 5 + 2` | `17` ✓ |

#### 4) กับดักที่เจอบ่อย
- สับสน `//` กับ `/` (ตัวเลือก C)
- สับสน `//` (ผลหาร) กับ `%` (เศษ)
- คิดว่า `-7 // 2` = `-3`: ที่จริง `-4` เพราะปัดลง

## ข้อ 4 — ⭐ กับดัก: `while True` + `break` ที่ย่อหน้าผิดและ `+ =`

What will be the output of the following Python code? *(ดูทุกตัวอักษรและทุกช่องว่างให้ดี)*

````python
    i = 1
while True:
    if i%3 == 0:
        break
    print(i)
    i + = 1
````

- A. 1 2 3
- B. error
- C. 1 2
- D. Done
- E. None of the above

### 📝 คำตอบของฉัน (ห้ามรันโค้ดก่อนตอบ!)
**ดับเบิลคลิกที่ cell นี้เพื่อพิมพ์**

- คำตอบ:
- ตัดตัวเลือกไหนทิ้ง เพราะอะไร:

<details>
<summary>✅ เฉลยข้อ 4 (เปิดเมื่อครูบอก)</summary>

**คำตอบ: B. error**

มี**กับดัก** 2 จุด:
1. บรรทัดแรก `    i = 1` ย่อหน้าทั้งที่ไม่ได้อยู่ในบล็อกไหน → `IndentationError: unexpected indent`
2. `i + = 1` มีช่องว่างคั่นกลาง `+=` → `SyntaxError`

> ถ้าเป็นแค่พิมพ์ผิด (แก้แล้ว) จะได้ **C. 1 2**: `i=1` พิมพ์ 1 · `i=2` พิมพ์ 2 · `i=3` → `3 % 3 == 0` → `break` ก่อนพิมพ์

- A ✘ `break` ทำงานก่อน `print(3)` · D ✘ ไม่มีคำว่า Done ในโค้ด

</details>

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
`while True:` คือลูปที่**ไม่มีวันหยุดเอง** ต้องใช้ `break` เพื่อออก ส่วนข้อนี้ยังทดสอบว่าอ่านโค้ดละเอียดพอจะเห็นกับดักไวยากรณ์หรือไม่

#### 2) กฎที่ต้องจำ
- `while True:` วนตลอด จนกว่าจะเจอ `break`
- `break` = ออกจากลูป**ทันที** บรรทัดที่เหลือในรอบนั้นไม่ทำ
- ลำดับในลูปสำคัญ: `if ... break` อยู่**ก่อน** `print` → ค่าที่ทำให้ break จะไม่ถูกพิมพ์
- บรรทัดที่ไม่ได้อยู่ในบล็อก (`if`, `while`, `def` ...) ต้อง**ชิดซ้าย** ย่อหน้าเกิน → `IndentationError`
- ตัวดำเนินการสองตัวอักษร (`+=`, `==`, `//`, `**`) ห้ามมีช่องว่างคั่นกลาง

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
โค้ดที่แก้แล้ว:

| รอบ | `i` | `i%3 == 0` | ทำอะไร |
|---|---|---|---|
| 1 | 1 | False | พิมพ์ 1, `i` → 2 |
| 2 | 2 | False | พิมพ์ 2, `i` → 3 |
| 3 | 3 | **True** | `break` ออกจากลูป (ไม่พิมพ์ 3) |

#### 4) กับดักที่เจอบ่อย
- ไม่เห็นช่องว่างหน้า `i = 1` และใน `+ =`
- คิดว่า `break` ทำหลัง `print` → ตอบ 1 2 3
- คิดว่า `while True` ต้องเป็น infinite loop เสมอ: มี `break` ก็ออกได้

## ข้อ 7 — ⭐ `filter(bool, ...)` เก็บเฉพาะค่าที่เป็นจริง

What is the expected output of the following code?

````python
l = [1,0,2,0,"hello", "", []]
print(list(filter(bool, l)))
````

- A. [1, 0, 2, 'hello', '', []]
- B. [1, 2, 'hello']
- C. [1, 0, 2, 0, 'hello', '', []]
- D. Error
- E. None of the above

### 📝 คำตอบของฉัน (ห้ามรันโค้ดก่อนตอบ!)
**ดับเบิลคลิกที่ cell นี้เพื่อพิมพ์**

- ตัดตัวเลือกไหนทิ้ง เพราะอะไร:

<details>
<summary>✅ เฉลยข้อ 7 (เปิดเมื่อครูบอก)</summary>

**คำตอบ: B. [1, 2, 'hello']**

`filter(bool, l)` เก็บเฉพาะตัวที่ `bool(ตัวนั้น)` เป็น `True`: `0`, `""`, `[]` เป็นเท็จจึงถูกทิ้ง

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
`filter(ฟังก์ชัน, ข้อมูล)` = **กรอง**: ส่งสมาชิกทีละตัวเข้าฟังก์ชัน เก็บเฉพาะตัวที่ได้ผลเป็นจริง
เมื่อใช้ `bool` เป็นฟังก์ชัน คือ "เก็บเฉพาะค่าที่ไม่ว่าง/ไม่เป็นศูนย์"

#### 2) กฎที่ต้องจำ
- ค่าที่เป็นเท็จ (falsy): `False`, `0`, `0.0`, `""`, `[]`, `()`, `{}`, `None`
- ค่าอื่นเป็นจริงหมด รวมถึง `"0"`, `" "`, `[0]`, `-1`
- `filter` คืนค่าเป็น filter object ต้องใส่ `list(...)` ถึงจะเห็นสมาชิก
- เขียนแบบ list comprehension ได้: `[x for x in l if x]`

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
| สมาชิก | `bool(สมาชิก)` | เก็บไหม |
|---|---|---|
| `1` | True | ✔ |
| `0` | False | ✘ |
| `2` | True | ✔ |
| `0` | False | ✘ |
| `"hello"` | True | ✔ |
| `""` | False | ✘ |
| `[]` | False | ✘ |

ได้ `[1, 2, 'hello']`

## ข้อ 9 — ⭐ กับดัก: `def__init__` ไม่มีช่องว่าง + `self.id` กับ `id`

What will be the output of the following Python code? *(ดูทุกตัวอักษรให้ดี)*

````python
class tester:
    def__init__(self, id):
        self.id = str(id)
        id = "224"

temp = tester(12)
print(temp.id)
````

- A. 12
- B. 224
- C. Error
- D. None
- E. None of the above

### 📝 คำตอบของฉัน (ห้ามรันโค้ดก่อนตอบ!)
**ดับเบิลคลิกที่ cell นี้เพื่อพิมพ์**

- คำตอบ:
- ตัดตัวเลือกไหนทิ้ง เพราะอะไร:

<details>
<summary>✅ เฉลยข้อ 9 (เปิดเมื่อครูบอก)</summary>

**คำตอบ: C. Error**

**กับดัก**: `def__init__` ไม่มีช่องว่างหลัง `def` → Python อ่านเป็นชื่อ `def__init__` ซึ่งไม่ใช่การประกาศฟังก์ชัน → `SyntaxError`

> ถ้าเป็นแค่พิมพ์ผิด (มีช่องว่าง) จะได้ **A. 12**: `self.id = str(12)` = `'12'` ส่วน `id = "224"` เปลี่ยนแค่ตัวแปร `id` ในฟังก์ชัน ไม่ได้เปลี่ยน `self.id`

- B ✘ (224) คิดว่าเปลี่ยน `id` แล้ว `self.id` เปลี่ยนตาม · D ✘ `self.id` มีค่าแล้ว

</details>

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
ใน method มีของสองอย่างที่ชื่อคล้ายกัน:
- `self.id` = ค่าที่**ติดอยู่กับ object** อยู่ต่อไปหลังจาก method จบ
- `id` = ตัวแปร (parameter) ของ method หายไปเมื่อ method จบ

เปลี่ยนตัวหนึ่ง ไม่ได้เปลี่ยนอีกตัว

#### 2) กฎที่ต้องจำ
- `def` ต้องตามด้วยช่องว่าง แล้วจึงเป็นชื่อฟังก์ชัน
- `self.id = str(id)` **คัดลอกค่า**ตอนนั้นไปเก็บ: เปลี่ยน `id` ทีหลังไม่กระทบ `self.id`
- `str(12)` → `'12'` (string) `print` จึงแสดง `12` เหมือนกัน
- `id` เป็นชื่อฟังก์ชัน built-in ด้วย ตั้งชื่อตัวแปรทับได้แต่ไม่ควร

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
โค้ดที่แก้แล้ว:

| ขั้น | ทำอะไร | `id` (ใน method) | `self.id` |
|---|---|---|---|
| `tester(12)` | เรียก `__init__` | `12` | — |
| `self.id = str(id)` | | `12` | `'12'` |
| `id = "224"` | | `'224'` | `'12'` (ไม่เปลี่ยน) |
| `print(temp.id)` | | | พิมพ์ `12` |

#### 4) กับดักที่เจอบ่อย
- ไม่เห็นว่า `def__init__` ติดกัน
- คิดว่า `self.id` กับ `id` เป็นตัวเดียวกัน (ตัวเลือก B)
- ลำดับบรรทัดสำคัญ: ถ้า `id = "224"` อยู่**ก่อน** `self.id = str(id)` คำตอบจะเป็น 224

ถ้าไม่ใช่แบบนี้ เราจะตั้งตัวแปรชื่อ `len` ในฟังก์ชันแล้วใช้ไม่ได้ เพราะ built-in จะถูกหาเจอก่อน

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
**namespace** คือ "สมุดรายชื่อ" ที่จับคู่ชื่อกับค่า เวลา Python เจอชื่อ มันเปิดสมุดจาก**ใกล้ตัวที่สุดก่อน** เจอเล่มไหนก่อนใช้เล่มนั้น เรียกว่ากฎ **LEGB**

#### 2) กฎที่ต้องจำ
1. **L**ocal: ในฟังก์ชันที่กำลังทำงาน
2. **E**nclosing: ในฟังก์ชันที่ครอบอยู่ (กรณีฟังก์ชันซ้อน)
3. **G**lobal: ระดับไฟล์ (นอกฟังก์ชันทุกตัว)
4. **B**uilt-in: `print`, `len`, `max` ...

- หาไม่เจอสักเล่ม → `NameError`
- ถ้าในฟังก์ชันมีการ**กำหนดค่า** `x = ...` Python ถือว่า `x` เป็น local ทั้งฟังก์ชัน
- อยากเปลี่ยนตัวแปร global ในฟังก์ชัน ต้องใช้ `global x`

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด

|---|---|---|
| `print(x)` ใน `f` | local มี `x` | `local` |
| `print(x)` นอก `f` | global | `global` |

- คิดว่า built-in มาก่อน (ตัวเลือก A, B)
- คิดว่าการตั้ง `x` ในฟังก์ชันไปเปลี่ยน `x` ข้างนอก

## ข้อ 11 — ⭐ กับดัก: `*args` / `**kwargs` และ `def` ที่ลืม `:`

The following python program can work with ___ parameters. *(ดูทุกตัวอักษรให้ดี)*

````python
def f(x)
    def f1(*args, **kwargs):
        print("Sanfoundry")
        return x(*args, **kwargs)
    return f1
````

- A. any number of
- B. 0
- C. 1
- D. 2
- E. None of the above

### 📝 คำตอบของฉัน (ห้ามรันโค้ดก่อนตอบ!)
**ดับเบิลคลิกที่ cell นี้เพื่อพิมพ์**

- คำตอบ:
- ตัดตัวเลือกไหนทิ้ง เพราะอะไร:

<details>
<summary>✅ เฉลยข้อ 11 (เปิดเมื่อครูบอก)</summary>

**คำตอบ: E. None of the above** (โค้ดตามที่พิมพ์รันไม่ได้เลย)

**กับดัก**: `def f(x)` ไม่มี `:` → `SyntaxError` โปรแกรมจึงทำงานกับพารามิเตอร์กี่ตัวก็ไม่ได้

> ถ้าเป็นแค่พิมพ์ผิด (มี `:`) จะได้ **A. any number of**: `f1(*args, **kwargs)` รับ argument กี่ตัวก็ได้ ทั้งแบบเรียงตำแหน่งและแบบมีชื่อ แล้วส่งต่อให้ `x` ทั้งหมด

</details>

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
`*args` และ `**kwargs` ทำให้ฟังก์ชัน**รับ argument กี่ตัวก็ได้**
ฟังก์ชัน `f` ในข้อนี้คือ **decorator**: รับฟังก์ชัน `x` เข้ามา แล้วคืนฟังก์ชันใหม่ `f1` ที่พิมพ์ข้อความก่อน แล้วค่อยเรียก `x` ด้วย argument ทั้งหมดที่ได้รับ

#### 2) กฎที่ต้องจำ
- `*args` เก็บ argument แบบเรียงตำแหน่งทั้งหมดเป็น **tuple**: `f1(1, 2)` → `args = (1, 2)`
- `**kwargs` เก็บ argument แบบมีชื่อทั้งหมดเป็น **dict**: `f1(a=1)` → `kwargs = {'a': 1}`
- `x(*args, **kwargs)` = แตกของในกล่องส่งต่อให้ `x` ครบทุกตัว
- ชื่อ `args`, `kwargs` ตั้งอย่างอื่นได้ ที่สำคัญคือ `*` กับ `**`
- `def` ต้องจบด้วย `:` เสมอ

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
โค้ดที่แก้แล้ว + การใช้งาน:

```python
def add(a, b):
    return a + b

g = f(add)
print(g(2, 3))
```

| ขั้น | ทำอะไร | ได้ |
|---|---|---|
| `f(add)` | สร้าง `f1` ที่จำ `x = add` ไว้ | `g` คือ `f1` |
| `g(2, 3)` | `args = (2, 3)`, `kwargs = {}` | พิมพ์ `Sanfoundry` |
| `x(*args)` | = `add(2, 3)` | `5` |
| `print(...)` | | พิมพ์ `5` |

#### 4) กับดักที่เจอบ่อย
- ไม่เห็นว่า `def f(x)` ไม่มี `:`
- คิดว่ารับได้แค่ 1 ตัว เพราะ `f(x)` มีพารามิเตอร์เดียว: แต่คำถามหมายถึง `f1` ที่ `f` คืนออกมา

## ข้อ 13 — ⭐ กับดัก: `[::-1]` + `end=''` ไม่มีช่องว่าง

What will be the output of the following Python code snippet? *(ดูทุกตัวอักษรให้ดี)*

````python
for i in [1, 2, 3, 4][::-1]:
    print(i, end='')
````

- A. 1 2 3 4
- B. 1 2 3
- C. 4 3 2 1
- D. 4 3 2
- E. None of the above

### 📝 คำตอบของฉัน (ห้ามรันโค้ดก่อนตอบ!)
**ดับเบิลคลิกที่ cell นี้เพื่อพิมพ์**

- คำตอบ:
- ตัดตัวเลือกไหนทิ้ง เพราะอะไร:

<details>
<summary>✅ เฉลยข้อ 13 (เปิดเมื่อครูบอก)</summary>

**คำตอบ: E. None of the above** (พิมพ์ `4321` ติดกัน)

`[::-1]` กลับลำดับ list → `[4, 3, 2, 1]` ส่วน `end=''` ทำให้ `print` ไม่ขึ้นบรรทัดใหม่**และไม่เว้นวรรค** ผลจึงเป็น `4321` ติดกัน

> ถ้าตัวเลือกในข้อสอบแค่เว้นวรรคให้อ่านง่าย คำตอบที่ตั้งใจคือ **C. 4 3 2 1**

- A ✘ ไม่ได้กลับลำดับ · B, D ✘ ไม่ได้ตัดตัวไหนทิ้ง

</details>

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
ข้อนี้รวม 2 เรื่อง: **slicing กลับลำดับ** (`[::-1]`) และ**การควบคุมสิ่งที่ `print` พิมพ์ต่อท้าย** (`end`)

#### 2) กฎที่ต้องจำ
- `[start:stop:step]` · `step = -1` = เดินถอยหลัง → `[::-1]` คือกลับลำดับทั้งหมด
- `print(x)` ปกติพิมพ์ `x` แล้ว**ขึ้นบรรทัดใหม่** (`end='\n'`)
- `end=''` = ไม่พิมพ์อะไรต่อท้าย · `end=' '` = เว้นวรรค · `end='-'` = ขีด
- `sep` คือตัวคั่นระหว่างหลายค่าใน `print` เดียว: `print(1, 2, sep=', ')` → `1, 2`

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
| รอบ | `i` | พิมพ์ | บรรทัดตอนนี้ |
|---|---|---|---|
| 1 | 4 | `4` | `4` |
| 2 | 3 | `3` | `43` |
| 3 | 2 | `2` | `432` |
| 4 | 1 | `1` | `4321` |

#### 4) กับดักที่เจอบ่อย
- คิดว่า `end=''` ยังเว้นวรรคให้
- สับสนว่า `[::-1]` ตัดตัวสุดท้ายทิ้ง (ตัวเลือก D)

#### 🧠 เช็คตัวเอง
ทายผล

```python

    print(c, end='-')
print()
print(1, 2, 3, sep='')
```


`end='-'` ต่อขีดหลังทุกตัวรวมตัวสุดท้าย · `sep=''` ทำให้สามค่าติดกัน

## ข้อ 14 — ⭐ ส่ง list เข้าฟังก์ชัน: object ตัวเดียวกัน (`id`)

What will be the output of the following Python program?

````python
def foo(x):
    x[0] = ['def']
    x[1] = ['abc']
    return id(x)

q = ['abc', 'def']
print(id(q) == foo(q))
````

- A. Error
- B. None

**คำตอบ: C. True**

ส่ง `q` เข้า `foo` → `x` ชี้ไปที่ **list ตัวเดียวกัน** กับ `q` การแก้ `x[0]`, `x[1]` คือการแก้ข้างใน list เดิม ไม่ได้สร้าง list ใหม่ → `id(x)` เท่ากับ `id(q)`
- D ✘ คิดว่าฟังก์ชันได้สำเนาใหม่ · A ✘ ใส่ list ซ้อน list ได้ปกติ

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
ตัวแปรใน Python เป็นเหมือน**ป้ายชื่อ**ที่แปะอยู่บนของ ไม่ใช่กล่องเก็บของ ส่ง list เข้าฟังก์ชันจึงเป็นการแปะป้ายใหม่ (`x`) บน list **ใบเดิม**
`id(obj)` บอกหมายเลขประจำตัวของ object: เลขเท่ากัน = ของชิ้นเดียวกัน

#### 2) กฎที่ต้องจำ
- `x[0] = ...`, `x.append(...)` = **แก้ของเดิม** (ทุกป้ายที่แปะอยู่เห็นการเปลี่ยนแปลง)
- `x = [...]` = **ย้ายป้าย** `x` ไปแปะของใหม่ ของเดิมไม่เปลี่ยน
- `a is b` ⇔ `id(a) == id(b)`

- tuple แก้ข้างในไม่ได้ (`TypeError`)

## ข้อ 15 — Python ทั้ง compile และ interpret

Is Python code compiled or interpreted?

- A. Python code is both compiled and interpreted
- B. Python code is neither compiled nor interpreted
- C. Python code is only compiled
- D. Python code is only interpreted
- E. None of the above

### 📝 คำตอบของฉัน (ห้ามรันโค้ดก่อนตอบ!)
**ดับเบิลคลิกที่ cell นี้เพื่อพิมพ์**

- คำตอบ:
- ตัดตัวเลือกไหนทิ้ง เพราะอะไร:

<details>
<summary>✅ เฉลยข้อ 15 (เปิดเมื่อครูบอก)</summary>

**คำตอบ: A**

Python **compile** โค้ดทั้งไฟล์เป็น **bytecode** ก่อน (จึงเจอ SyntaxError ก่อนรันสักบรรทัด) แล้ว **interpret** bytecode ทีละคำสั่ง (จึงเกิด error ตอนรันได้หลังจากพิมพ์อะไรไปแล้ว)

</details>

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
เวลารันโค้ด Python มี 2 ขั้น:
1. **compile**: แปลงทั้งไฟล์เป็น bytecode (ภาษากลางที่เครื่องอ่านง่าย) ตรงนี้ตรวจไวยากรณ์
2. **interpret**: ตัวแปลภาษา (Python virtual machine) ทำ bytecode ทีละคำสั่ง

#### 2) กฎที่ต้องจำ
- **SyntaxError** เกิดตอน compile → ไม่มีบรรทัดไหนทำงานเลย
- **error อื่น** (`ZeroDivisionError`, `NameError` ...) เกิดตอน interpret → บรรทัดก่อนหน้าทำงานไปแล้ว
- `compile(ข้อความ, ชื่อ, 'exec')` สร้าง code object · `exec(code)` รันมัน
- `eval(ข้อความ)` คำนวณนิพจน์แล้วคืนค่า

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
```python
print("ก่อน")
print(1 / 0)
print("หลัง")
```

| ขั้น | เกิดอะไร |
|---|---|
| compile | ไวยากรณ์ถูกทั้งหมด → ผ่าน |
| interpret บรรทัด 1 | พิมพ์ `ก่อน` |
| interpret บรรทัด 2 | `ZeroDivisionError` → หยุด |
| บรรทัด 3 | ไม่ได้ทำ |

#### 4) กับดักที่เจอบ่อย
- คิดว่า Python เป็นแค่ interpreted (ตัวเลือก D) แต่ SyntaxError ที่ท้ายไฟล์ทำให้บรรทัดแรกไม่ทำงาน แปลว่ามีการ compile ทั้งไฟล์ก่อน

#### 🧠 เช็คตัวเอง
ทายผล

```python
src = 'print("A")\nprint("B"'
try:
    exec(src)
except SyntaxError:
    print('SyntaxError ก่อนพิมพ์ A')
```

<details>
<summary>เฉลยเช็คตัวเอง</summary>


### 📝 คำตอบของฉัน (ห้ามรันโค้ดก่อนตอบ!)
**ดับเบิลคลิกที่ cell นี้เพื่อพิมพ์**

**คำตอบ: D. A, B, C, D บรรทัดละตัว**

วน `'abcd'` ได้ทีละตัวอักษร → `.upper()` ทำเป็นตัวใหญ่ → `print` แต่ละครั้ง**ขึ้นบรรทัดใหม่**
- C ✘ อยู่บรรทัดเดียว (ต้องใช้ `end=' '`) · A, B ✘ ตัว a ไม่ได้รอดจาก `.upper()`

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
string วนด้วย `for` ได้ทีละตัวอักษร และ method ของ string อย่าง `.upper()` สร้าง string ใหม่ที่เป็นตัวใหญ่

#### 2) กฎที่ต้องจำ
- `for ch in 'abcd':` → `ch` เป็น `'a'`, `'b'`, `'c'`, `'d'` ตามลำดับ
- `.upper()` / `.lower()` คืน string **ใหม่** ของเดิมไม่เปลี่ยน
- ตัวเลขหรือเครื่องหมายไม่เปลี่ยน: `'a1!'.upper()` → `'A1!'`
- ต้องการเก็บผล ต้องกำหนดค่าใหม่: `x = x.upper()`

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
| รอบ | `i` | `i.upper()` | output |
|---|---|---|---|
| 1 | `'a'` | `'A'` | `A` (ขึ้นบรรทัดใหม่) |
| 2 | `'b'` | `'B'` | `B` |
| 3 | `'c'` | `'C'` | `C` |
| 4 | `'d'` | `'D'` | `D` |

#### 4) กับดักที่เจอบ่อย
- คิดว่าพิมพ์อยู่บรรทัดเดียว (ตัวเลือก C)
- คิดว่า `x.upper()` เปลี่ยน `x` เอง

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
`bool` เป็นตัวเลขชนิดพิเศษ: `False` เท่ากับ `0` และ `True` เท่ากับ `1` จึงเปรียบเทียบกับตัวเลขได้ แต่ `max`/`min` คืน**ตัวเดิม** ไม่ได้แปลงเป็นตัวเลข

#### 2) กฎที่ต้องจำ
- `False == 0`, `True == 1`, `True + True == 2`
- `max(...)` / `min(...)` คืนสมาชิกตัวที่มาก/น้อยที่สุด **ตามที่ส่งไป** (ถ้าเท่ากันคืนตัวแรก)

| ขั้น | ทำอะไร | ได้ |
|---|---|---|
| 1 | `max(False, -3, -4)` เทียบ `0, -3, -4` | `False` |
| 2 | `min(False, 2, 7)` เทียบ `0, 2, 7` | `False` |
| 3 | `print(False)` | `False` |

#### 4) กับดักที่เจอบ่อย
- คิดว่า `max` ข้าม `False` ไป (ตัวเลือก D)
- คิดว่าผลจะเป็น `0` แทน `False`

> `<<` คือเลื่อนบิตไปทางซ้าย = คูณ 2 ยกกำลังจำนวนที่เลื่อน: `x << 2` = `x * 4` ถ้า `x = 1` ได้ **C. 4** · ถ้า `x = 2` ได้ **D. 8**

</details>

### 📚 ความรู้ที่ต้องรู้

#### 1) แนวคิด
ตัวเลขในคอมพิวเตอร์เก็บเป็นเลขฐานสอง `<<` คือเลื่อนบิตทั้งหมดไปทางซ้าย ซึ่งเท่ากับ**คูณด้วย 2** ทุกครั้งที่เลื่อน 1 ช่อง ส่วน `>>` เลื่อนไปทางขวา = หาร 2 ปัดลง

#### 2) กฎที่ต้องจำ
- `x << n` = `x * 2**n` · `x >> n` = `x // 2**n`
- `bin(5)` → `'0b101'` ดูเลขฐานสองได้
- ใช้ตัวแปรที่ยังไม่เคยกำหนดค่า → `NameError`

#### 3) ตัวอย่าง: ไล่ทีละบรรทัด
```python
x = 1
print(x << 2)
```

| ขั้น | ฐานสอง | ฐานสิบ |
|---|---|---|
| `x` | `0b1` | 1 |
| เลื่อนซ้าย 1 | `0b10` | 2 |
| เลื่อนซ้าย 2 | `0b100` | **4** |

#### 4) กับดักที่เจอบ่อย
- ตอบเลขทันทีโดยไม่ดูว่า `x` ไม่มีค่า
- สับสน `<<` (คูณ) กับ `>>` (หาร)

#### 🧠 เช็คตัวเอง
ทายผล

```python
print(3 << 1, 3 << 3, 20 >> 2)
```

<summary>เฉลยเช็คตัวเอง</summary>

```
6 24 5
```

`3 * 2 = 6` · `3 * 8 = 24` · `20 // 4 = 5`
